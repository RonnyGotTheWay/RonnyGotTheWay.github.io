```python
import pandas as pd
import numpy as np

# 假设数据已经加载到 `data_cleaned` 数据框中
def calculate_avg_scores_before_elimination(df):
    avg_scores = []  # 存储所有选手的平均分

    # 遍历所有选手
    for index, row in df.iterrows():
        celebrity_name = row['celebrity_name']
        scores = []  # 用于存储每周的评分
        eliminated = False  # 用于标记选手是否被淘汰
        
        # 遍历每周的评分（最多11周）
        for week in range(1, 12):  # 11周的数据
            week_scores = []
            for judge in range(1, 5):  # 每周有4个评委评分
                score_column = f'week{week}_judge{judge}_score'
                
                if pd.notna(row[score_column]) and row[score_column] != 0:  # 如果该评分不是NaN且不为0
                    week_scores.append(row[score_column])
                
            if week_scores:  # 如果存在评分
                # 计算每周的评分平均值
                scores.append(np.mean(week_scores))
            
            # 如果某一周存在评委评分为0，标记为被淘汰
            if 0 in week_scores:
                eliminated = True
                break
        
        # 如果被淘汰，计算淘汰前的平均分
        if eliminated:
            avg_scores.append(np.mean(scores))  # 计算被淘汰前的评分平均值
        else:
            avg_scores.append(np.mean(scores))  # 未淘汰的选手计算平均值
    
    df['avg_scores_before_elimination'] = avg_scores
    return df

# 假设 `data_cleaned` 是你已经清理过的数据
# 调用函数计算每个选手的记忆项（被淘汰前的平均评分）
data_cleaned = calculate_avg_scores_before_elimination(data_cleaned)

# 检查生成的记忆项
data_cleaned[['celebrity_name', 'avg_scores_before_elimination']].head(20)
```

```python
import pandas as pd

# Load your cleaned data
data = pd.read_csv('cleaned_data_with_name.csv')  # Replace with your actual file path

# List of columns to delete
columns_to_delete = [
    "week1_total_score", "week1_avg_score", "week2_total_score", "week2_avg_score",
    "week3_total_score", "week3_avg_score", "week4_total_score", "week4_avg_score",
    "week5_total_score", "week5_avg_score", "week6_total_score", "week6_avg_score",
    "week7_total_score", "week7_avg_score", "week8_total_score", "week8_avg_score",
    "week9_total_score", "week9_avg_score", "week10_total_score", "week10_avg_score",
    "week11_total_score", "week11_avg_score","results_encoded"
]

# Drop the specified columns
data_cleaned = data.drop(columns=columns_to_delete)

import numpy as np
import pandas as pd

def add_avg_score_up_to_t(df, t, weeks_max=11, judges_max=4,
                          out_col="avg_score_up_to_t", treat_zero_as_missing=True):
    """
    计算每个选手“前 t 周的平均分”（先算每周评委均分，再对 1..t 周取平均）。
    - df: 宽表（包含 week{w}_judge{j}_score）
    - t: 当前周（实时预测时你知道当前是第几周）
    """
    t = int(t)
    t = max(1, min(t, weeks_max))

    weekly_mean_list = []
    for week in range(1, t + 1):
        judge_cols = [
            f"week{week}_judge{j}_score"
            for j in range(1, judges_max + 1)
            if f"week{week}_judge{j}_score" in df.columns
        ]
        if not judge_cols:
            continue

        scores = df[judge_cols].copy()
        if treat_zero_as_missing:
            scores = scores.replace(0, np.nan)

        # 该周评委均分（每行一个选手）
        weekly_mean_list.append(scores.mean(axis=1, skipna=True))

    if not weekly_mean_list:
        df[out_col] = np.nan
        return df

    # 前 t 周均分：对每周均分再取平均
    df[out_col] = pd.concat(weekly_mean_list, axis=1).mean(axis=1, skipna=True)
    return df

# ===== 用法示例 =====
current_week = 6  # 你预测时处于第几周，就填几（实时预测场景一定知道）
data_cleaned = add_avg_score_up_to_t(data_cleaned, t=current_week, out_col="avg_score_up_to_t")

# 看一下结果
data_cleaned[["celebrity_name", "avg_score_up_to_t"]].head(10)


# Save the cleaned data to a new file
data_cleaned.to_csv('cleaned_data_2.csv', index=False)
```

```python
import pandas as pd
import numpy as np

# 读取数据（确保文件路径正确）
data = pd.read_csv('data_with_partner_strength.csv')

# 计算每一周评委评分的方差
def calculate_weekly_variance(df):
    # 为每一周的方差创建一个新的列
    for week in range(1, 12):  # 遍历11周
        week_scores = []  # 用于存储该周所有选手的方差
        
        # 遍历每个选手
        for index, row in df.iterrows():
            week_column_scores = []
            for judge in range(1, 5):  # 每周有4个评委评分
                score_column = f'week{week}_judge{judge}_score'
                
                if pd.notna(row[score_column]) and row[score_column] != 0:  # 如果该评分不是NaN且不为0
                    week_column_scores.append(row[score_column])
            
            # 如果该选手该周有评分，计算方差，否则填充NaN
            if week_column_scores:  # 如果有有效评分
                week_scores.append(np.var(week_column_scores))  # 每个选手该周的方差
            else:
                week_scores.append(np.nan)  # 如果没有评分，填充NaN
        
        # 确保列长度与原数据框一致
        if len(week_scores) < len(df):
            week_scores.extend([np.nan] * (len(df) - len(week_scores)))
        
        # 将计算的方差作为新的列加入数据框
        df[f'week{week}_variance'] = week_scores
    
    return df

# 调用函数计算每个选手每周的评分方差
data = calculate_weekly_variance(data)

# 检查生成的方差列
print(data[['celebrity_name', 'week1_variance', 'week2_variance', 'week3_variance', 'week4_variance', 'week5_variance', 'week6_variance', 'week7_variance', 'week8_variance', 'week9_variance', 'week10_variance', 'week11_variance']].head(20))
```

```python
import pandas as pd

# 读取更新后的数据
data = pd.read_csv('data_with_variance_added.csv')

# 设置方差阈值
threshold = 0.68

# 列出所有 weekX_variance 列（总共11周，每周4个评委）
variance_columns = [
    'week1_variance', 'week2_variance', 'week3_variance', 'week4_variance', 'week5_variance',
    'week6_variance', 'week7_variance', 'week8_variance', 'week9_variance', 'week10_variance', 'week11_variance'
]

# 为每个 weekX_variance 列生成争议标签
for variance_column in variance_columns:
    controversy_label = []
    
    for index, row in data.iterrows():
        variance = row[variance_column]
        
        # 如果方差大于阈值0.68，则标记为True，否则为False
        if pd.notna(variance) and variance > threshold:
            controversy_label.append(True)
        else:
            controversy_label.append(False)
    
    # 将争议标签添加到数据框中
    data[f'{variance_column}_contr'] = controversy_label

# 保存更新后的数据框
data.to_csv('data_with_controversy_labels.csv', index=False)
data.to_csv(r"C:\Users\ronny\Desktop\美赛\data\data_with_controversy_labels.csv", index=False)

# 查看最终结果（显示前20行）
print(data[['celebrity_name'] + [f'{col}_contr' for col in variance_columns]].head(20))
```
